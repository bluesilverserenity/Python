```python
import cv2
import numpy as np

im_g = cv2.imread("smallgray.png",0)
print(im_g)

im_g[0:2,2:4]
```

```python
for i in im_g:  
    print(i)   #prints each row of the image as a 1D array

for i in im_g.T:
    print(i)    #prints each column of the image as a 1D array 
```

```python
sna = np.hstack([im_g,im_g]) #stacks the image horizontally (side by side)
print(sna)

sna2 = np.vstack([im_g,im_g]) #stacks the image vertically (one on top of the other)
print(sna2)
```

```python
sla = np.hsplit(sna,2) #splits the horizontally stacked image into 2 equal parts
print(sla)

sla2 = np.vsplit(sna,3) #splits the vertically stacked image into 3 equal parts
print(sla2)
```
