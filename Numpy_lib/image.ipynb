```python
import cv2
```

```python
im_g = cv2.imread("smallgray.png",0) #reads the image in grayscale mode and stores it in the variable im_g
print(im_g) #prints the pixel values of the image as a 2D array

im_g2 = cv2.imread("smallgray.png",1) #reads the image in color(BGR) mode and stores it in the variable im_g2
print(im_g2) #prints the pixel values of the image as a 3D array
```

```python
cv2.imwrite("newsmallgray.png",im_g) #saves the grayscale image as a new file named "newsmallgray.png"
```
