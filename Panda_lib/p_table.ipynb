```python
import pandas as pd

df1 = pd.DataFrame([[3, 4], [20, 14]], columns=['Age', 'Price'])
print(df1)

df2 = pd.DataFrame([[3, 4], [20, 14], [25, 18]], columns=['Age', 'Price'])
print(df2[['Age']])
```

```python
import pandas as pd
l1 = [3, 20]
l2 = [4, 14]
df2 = pd.DataFrame(list(zip(l1, l2)), columns=['Age', 'Price'])
print(df2)

dl = [{'Name': 'Alice', 'Age': 14}, {'Name': 'Bob', 'Age': 16}]
df3 = pd.DataFrame(dl)
print(df3)

dl2 = {"Name" : ["Alice", "Bob"], "Age" : [14, 16]}
df4 = pd.DataFrame(dl2)
print(df4)
```
