```python
from geopy.geocoders import Nominatim
nom = Nominatim(user_agent="myGeocoder")
l = nom.geocode("175 5th Avenue NYC")
l.latitude
l.longitude #prints the latitude and longitude of the address
l
```

```python
import pandas as pd
df = pd.read_json("supermarkets.json")
df
```

```python
df['Address'] = df['Address'] + ',' + df['City'] + ',' + df['State'] + ',' + df['Country']
df
```

```python
df['Coordinates'] = df['Address'].apply(nom.geocode)
df['Lat&Long'] = df['Coordinates'].apply(lambda x: (x.latitude, x.longitude) if x else None) #creates a new column 'Lat&Long' that contains the latitude and longitude as a tuple for each address. If the geocoding fails (i.e., if x is None), it assigns None to that entry.
df
df.Coordinates[0] #prints the coordinates of the first address in the dataframe
```
